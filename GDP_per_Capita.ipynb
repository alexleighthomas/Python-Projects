```python
GDP_df['WorldBank_Estimate'] = pd.to_numeric(GDP_df['WorldBank_Estimate'], errors='coerce').fillna(0).astype(int)
```

```python
print(GDP_df.dtypes)
```

```python
col = 'IMF_Estimate'
Q1 = GDP_df[col].quantile(0.25)
Q3 = GDP_df[col].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = GDP_df[(GDP_df[col] < lower) | (GDP_df[col] > upper)]
```

```python
import matplotlib.pyplot as plt
numeric_cols = ['IMF_Estimate', 'WorldBank_Estimate', 'UN_Estimate']
data_to_plot = [GDP_df[col].dropna() for col in numeric_cols]

plt.figure(figsize=(8, 4))
plt.boxplot(data_to_plot, vert=False, labels=numeric_cols)
plt.title('Boxplot comparison (outliers shown as dots)')
plt.xlabel('Value')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print(student_df.columns.tolist())
```

```python
# To create a Histogram:
plt.figure(figsize=(10, 6))
plt.hist(student_df['mark'], bins=10, edgecolor='skyblue')
plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist(student_df['mark'], color='#ff5733', edgecolor='black')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
counts, bins, patches = plt.hist(student_df['mark'], bins=10, edgecolor='black')

colors = ['red', 'blue', 'green', 'orange', 'purple', 'cyan', 'yellow', 'pink', 'brown', 'grey']
for patch, color in zip(patches, colors):
    patch.set_facecolor(color)

plt.xlabel('Mark')
plt.ylabel('Frequency')
plt.title('Marks histogram with different colours per bin')
plt.show()
```
